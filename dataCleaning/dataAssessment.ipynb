```python
duplicate_values = df.duplicated().sum()
print("The number of duplicate values are : ", duplicate_values)
```

```python
# Lets handle the missing values in our Numeric and Categorical columns
# For numeric columns, we will fill the missing values with the median of the column
numeric_cols = df.select_dtypes(include='number').columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

#check if there are still any null values in the numeric columns
df[numeric_cols].isnull().sum()
```

```python
# For categorical  columns, we will fill the missing values with the most frequent value dominant in the column
categ_cols = df.select_dtypes(include= ['object', 'category']).columns
for col in categ_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
#check if there are still any null values in the categorical columns
df[categ_cols].isnull().sum()   
```

```python
# Check if there are any missing values left by displaying  the first  & last 10 rows of the DataFrame after handling missing values
print(df.head(10))    
print(df.tail(10))
```

```python
# Check for duplicate rows and count them
duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
    print(f"Duplicate rows found: {duplicate_rows}")
else:
    print(f"No duplicate rows found: {duplicate_rows} ows")
```

```python
# Check for unique values in each column
unique_values = {col: df[col].nunique() for col in df.columns}
# Display the number of unique values in each column
print("Number of unique values in each column:")
for col, count in unique_values.items():
    print(f"{col}: {count} unique values")
```

```python
# Convert date columns to datetime format
date_columns = ['issuanceDate', 'expiryDate', 'claimDate', 'interactionDate', 'renewalDate']
for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
df['expiryDate']
```

```python
# Check for outliers in numeric columns using the IQR method
outlier_thresholds = {}
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outlier_thresholds[col] = (lower_bound, upper_bound)
# Display the outlier thresholds for each numeric column
print("Outlier thresholds (lower, upper) for numeric columns:")
for col, (lower, upper) in outlier_thresholds.items():
    print(f"{col}: ({lower}, {upper})")
    
```

```python
# Check for outliers in numeric columns using boxplot

for col in df.select_dtypes(include=[np.number]).columns:
    plt.figure(figsize=(10, 5))
    df.boxplot(column=col)
    plt.title(f"Boxplot of {col}")
    plt.show()      
```

The boxplots above shows that there are outliers present in three variables or columns. These are the age, income and the claimAmount columns. While the first two mentioned are similar since they have very long tails at both ends signifying a tighter age and income distribution where many people  of similar age earn very similar income. The boxplot for the claimAmount column shows that most claims fall between 704 and 1,255 approx. The outliers (long tail of negative clusters) fall below the 25th percentile(Q1) and are negative. This is unusual for claims and should call for further investigation and research so as to ascertain what might be responsible (might be from refunds and adjustments, data entry errors, claim reversals in the dataset etc.).  

```python
# Check for correlations between numeric columns using the correlation matrix


correlation_matrix = df.corr(numeric_only=True)
# Display the correlation matrix
print("Correlation matrix:")
print(correlation_matrix)
```

```python
# Visualize the correlation matrix using a heatmap

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Matrix Heatmap") 

plt.show()  
```

###  Insights drawn from the Correlation Matrix
 - Most values in this heatmap are closer to 0. This sindicates that there weak or no linear correlation between most variables.
 - Note the perfect correlation between customerID and policyID. This is as a result of the direct mapping of a common column during the data extraction   phase from the database( As no new information is added from having these columns we will have to drop one of them).
 - The correlation of the variable age with everyother variable is low.
 - ResponseTime has no meaninful corelations with any variable.
 - No strong correlation between premium, claimAmount and income(they have correlations around 0.00 – 0.02, which means that Higher income does not predict higher claims or premiums. Also, premium and renewalPremium have very weak correlation (-0.06))
 - Linear correlation is very low here so it is not advisable to rely on these features for prediction.
